```python
# Create RFM dataset
valid_statuses = [
    status
    for status in ["Delivered", "Completed"]
    if status in df["order_status"].unique()
]

valid_statuses
```

```python
rfm_df = df[
    df["order_status"].isin(valid_statuses)
].copy()
```
