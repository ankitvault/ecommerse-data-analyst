```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"  # so that invalid dates become Nat instead of reasing an error
    
)
```

```python
df["order_date"].dtype
```

```python
df["order_date"].isnull().sum()
```

```python
# Step 7 — Clean text columns #

text_columns = [
    "order_id",
    "customer_id",
    "payment_method",
    "ship_mode",
    "order_status"
]
# .str.strip() this removes leading and trailing whitespace from the string values in the specified columns
for col in text_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
# Create date features 
#Create date features
#Now that order_date is a proper datetime column, create useful features.
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["month_name"] = df["order_date"].dt.month_name()
df["quarter"] = df["order_date"].dt.quarter
df["day"] = df["order_date"].dt.day

df["day_name"] = df["order_date"].dt.day_name()
```

```python
#Create Year-Month
#This one will be especially useful for our monthly revenue chart.
df["year_month"] = (
    df["order_date"]
    .dt.to_period("M")
    .astype(str)
)
```
